# Visualize gaze falling into bounding boxes (Time series only)

```python
from pathlib import Path
from tobii_pytracker.analyze.data_loader import DataLoader
from tobii_pytracker.analyze.models import BBoxTimeSeriesAnalyzer
from tobii_pytracker.configs.custom_config import CustomConfig

config = globals().get("config")
if config is None:
    config = CustomConfig("../configs/config.yaml")

loader = globals().get("loader")
if loader is None:
    loader = DataLoader(config, root="../")

set_name = loader.get_subjects()[-1]
slide_index = 0
slide_data = loader.get_slide_data(set_name, slide_index, flatten=True)

analyzer = BBoxTimeSeriesAnalyzer(output_folder=Path("../output"))
scored = analyzer.analyze(slide_data, set_name=set_name, slide_index=slide_index)
scored[["bbox_index", "channel_idx", "hit_count", "is_visited"]]

analyzer.plot_analysis(
    slide_data=slide_data,
    scored_bboxes=scored,
    set_name=set_name,
    slide_index=slide_index,
    area_x=config.get_area_of_interest_size()[0],
    area_y=config.get_area_of_interest_size()[1],
    show=True,
)
```

# Visualize gaze falling into bounding boxes (Images only)

```python
from pathlib import Path
from tobii_pytracker.analyze.data_loader import DataLoader
from tobii_pytracker.analyze.models import BBoxImageAnalyzer
from tobii_pytracker.configs.custom_config import CustomConfig

config = globals().get("config")
if config is None:
    config = CustomConfig("../configs/config.yaml")

loader = globals().get("loader")
if loader is None:
    loader = DataLoader(config, root="../")

set_name = loader.get_subjects()[-1]
slide_index = 0
slide_data = loader.get_slide_data(set_name, slide_index, flatten=True)

analyzer = BBoxImageAnalyzer(output_folder=Path("../output"))
scored = analyzer.analyze(slide_data, set_name=set_name, slide_index=slide_index)
scored[["bbox_index", "hit_count", "is_visited"]].head()

screenshot_rel = slide_data.iloc[0]["screenshot_file"]
screenshot_path = Path("../") / Path(str(screenshot_rel).replace("\\", "/"))

analyzer.plot_analysis(
    scored_bboxes=scored,
    gaze_data=slide_data,
    screenshot_path=screenshot_path,
    set_name=set_name,
    slide_index=slide_index,
    show=True,
)
```

# Visualize gaze falling into bounding boxes (Text only)

```python
from pathlib import Path
from tobii_pytracker.analyze.data_loader import DataLoader
from tobii_pytracker.analyze.models import BBoxTextAnalyzer
from tobii_pytracker.configs.custom_config import CustomConfig

config = globals().get("config")
if config is None:
    config = CustomConfig("../configs/config.yaml")

loader = globals().get("loader")
if loader is None:
    loader = DataLoader(config, root="../")

set_name = loader.get_subjects()[-1]
slide_index = 0
slide_data = loader.get_slide_data(set_name, slide_index, flatten=True)

analyzer = BBoxTextAnalyzer(output_folder=Path("../output"))
scored = analyzer.analyze(
    slide_data,
    level="words",
    set_name=set_name,
    slide_index=slide_index,
)
scored[["bbox_index", "text", "hit_count", "is_visited"]].head()

analyzer.plot_analysis(
    slide_data=slide_data,
    level="words",
    scored_bboxes=scored,
    set_name=set_name,
    slide_index=slide_index,
    area_x=config.get_area_of_interest_size()[0],
    area_y=config.get_area_of_interest_size()[1],
    show=True,
)
```
